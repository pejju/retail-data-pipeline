## 5. Flag Non-Product Codes
StockCodes like POST, D, M, DOT, etc. are administrative, not real products — investigated via Description, flagged.

```python
data[data['StockCode'].str.isalpha()]['StockCode'].unique()
```

```python
data[data['StockCode'].isin(['POST','D','M','m','PADS','ADJUST','DCGSSGIRL','GIFT','DCGSLBOY','DCGSSBOY','DCGSLGIRL','S','B','AMAZONFEE'])][['StockCode','Description']].drop_duplicates()
```

```python
non_product_code = ['POST', 'D', 'M', 'DOT', 'ADJUST', 'S', 'B', 'AMAZONFEE', 'GIFT']
```

```python
data['isNonProduct'] = data['StockCode'].isin(non_product_code)
```

```python
data['isNonProduct'].sum()
```

## 6. Fix Missing Descriptions
Recovered ~2,563 missing descriptions via StockCode lookup; filled remaining 365 unrecoverable ones with 'Unknown Product'.

```python
desc_counts = data.groupby('StockCode')['Description'].count()
```

```python
desc_counts[desc_counts==0]
```

```python
unrecoverable_codes = desc_counts[desc_counts == 0].index
data[data['StockCode'].isin(unrecoverable_codes)].shape[0]
```

```python
desc_map = data.dropna(subset=['Description']).groupby('StockCode')['Description'].first()
```

```python
data['Description'] =  data['Description'].fillna(data['StockCode'].map(desc_map))
```

```python
data['Description'].isnull().sum()
```

```python
data['Description'] = data['Description'].fillna('Unknown Product')
```

```python
data['Description'].isnull().sum()
```

## 7. Handle Invalid Prices
Found 3 extreme negative-price 'bad debt' entries (already non-product) and ~3,681 zero-price rows. Built combined IsValidSale flag.

```python
data['Price'].describe()
```

```python
data['Price']<=0
```

```python
data.columns
```

```python
data[data['Price'] <= 0][['Invoice','StockCode','Description','Quantity','Price','IsCancelled','isNonProduct']]
```

```python
data[data['Price'] < 0]
```

```python
data['IsValidSale'] = ~data['IsCancelled'] & ~data['isNonProduct'] & (data['Price'] > 0)
```
